```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")

text = "Siva is learning how LLMs convert text to numbers."
tokens = tokenizer.encode(text)
print(tokens)
print(tokenizer.decode(tokens))
```

```python
subword_tokens = tokenizer.convert_ids_to_tokens(tokens)
print(subword_tokens)
```

```python
tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-7B-Instruct")

messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is a tokenizer?"}
]

formatted = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
print(formatted)
```

```python
models_to_compare = [
    "Qwen/Qwen2.5-7B-Instruct",
    "microsoft/Phi-4",
    "deepseek-ai/deepseek-coder-1.3b-instruct"
]

text = "Tokenization is fascinating!"

for model_name in models_to_compare:
    tok = AutoTokenizer.from_pretrained(model_name)
    ids = tok.encode(text)
    print(f"{model_name}: {len(ids)} tokens -> {tok.convert_ids_to_tokens(ids)}")
```

```python
text2 = "def calculate_sum(a, b): return a + b  # 日本語"

for model_name in models_to_compare:
    tok = AutoTokenizer.from_pretrained(model_name)
    ids = tok.encode(text2)
    print(f"{model_name}: {len(ids)} tokens")
```
