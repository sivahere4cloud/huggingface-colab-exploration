```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch

model_name = "distilbert/distilbert-base-cased-distilled-squad"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name).to("cuda")

question = "What is Siva's career goal?"
context = "Siva is an IT Manager working toward becoming an AI Engineer in the UK. He is currently learning Hugging Face tools as part of his transition."

inputs = tokenizer(question, context, return_tensors="pt").to("cuda")

with torch.no_grad():
    outputs = model(**inputs)

answer_start = outputs.start_logits.argmax()
answer_end = outputs.end_logits.argmax() + 1

answer = tokenizer.decode(inputs["input_ids"][0][answer_start:answer_end])
print(answer)
```
